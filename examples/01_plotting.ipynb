# Plotting with Poniard

Poniard's plotting is a separate module that requires plotly: `pip install poniard[plot]`

```python
from sklearn.datasets import make_classification
from poniard import PoniardClassifier
from poniard.plot import PoniardPlotFactory

X, y = make_classification(n_samples=200, n_features=10, random_state=42)
clf = PoniardClassifier()
clf.fit(X, y)
```

```python
plotter = PoniardPlotFactory(X, y, clf)
```

```python
plotter.metrics()
```

```python
plotter.roc_curve()
```

```python
plotter.confusion_matrix("LogisticRegression")
```

```python
plotter.permutation_importance("LogisticRegression")
```
